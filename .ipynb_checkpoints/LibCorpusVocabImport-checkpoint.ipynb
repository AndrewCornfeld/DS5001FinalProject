```python
import pandas as pd
import numpy as np
import seaborn as sns
import nltk
import re
```

```python
csv_file = "the-office-lines - scripts.csv"
```

```python
OHCO = ['season', 'episode', 'scene', 'line', 'sentence', 'token']
```

```python
raw = pd.read_csv(csv_file)
```

```python
raw.rename(columns={'id': 'line'}, inplace=True)
cols = list(raw.columns)
cols.insert(3, cols.pop(cols.index('line')))
raw = raw[cols]
```

```python
#raw.index.name = 'line'
raw
```

```python
sent_pat = r'[.?!;:]+'
SENTS = raw['line_text'].str.split(sent_pat, expand=True).stack()\
    .to_frame('sentence')
    
cols_to_keep = ['season', 'episode', 'scene', 'line']  # Columns to keep
SENTS[cols_to_keep] = raw.loc[SENTS.index.get_level_values(0), cols_to_keep].values
SENTS.set_index(OHCO[:4], inplace=True)
SENTS
```

```python
sample_sent = SENTS.iloc[4].sentence
```

```python
print(sample_tagged_tokens)
```

```python
keep_whitespace = True
```

```python
if keep_whitespace:
    # Return a tokenized copy of text
    # using NLTK's recommended word tokenizer.
    TOKENS = SENTS.sentence\
            .apply(lambda x: pd.Series(nltk.pos_tag(nltk.word_tokenize(x))))\
            .stack()\
            .to_frame('pos_tuple')
else:
    # Tokenize a string on whitespace (space, tab, newline).
    # In general, users should use the string ``split()`` method instead.
    # Returns fewer tokens.
    TOKENS = SENTS.sentence\
            .apply(lambda x: pd.Series(nltk.pos_tag(nltk.WhitespaceTokenizer().tokenize(x))))\
            .stack()\
            .to_frame('pos_tuple')
```

```python
TOKENS.index.names = OHCO[:5]
TOKENS
```

```python
TOKENS['pos'] = TOKENS.pos_tuple.apply(lambda x: x[1])
TOKENS['token_str'] = TOKENS.pos_tuple.apply(lambda x: x[0])
TOKENS['term_str'] = TOKENS.token_str.str.lower().str.replace(r"\W+", "", regex=True)
```

```python
TOKENS['pos_group'] = TOKENS.pos.str[:2]
TOKENS.head()
```

```python
VOCAB = TOKENS.term_str.value_counts().to_frame('n')
VOCAB.index.name = 'term_str'
VOCAB['p'] = VOCAB.n / VOCAB.n.sum()
VOCAB['i'] = -np.log2(VOCAB.p)
VOCAB['n_chars'] = VOCAB.index.str.len()
```

```python
bags = dict(
    SENTS = OHCO[:5],
    LINES = OHCO[:4],
    SCENES = OHCO[:3],
    EPISODES = OHCO[:2],
    SEASONS = OHCO[:1]
)
bag = 'LINES'
```

```python
if 'term_rank' not in VOCAB.columns:
    VOCAB = VOCAB.sort_values('n', ascending=False).reset_index()
    VOCAB.index.name = 'term_rank' 
    VOCAB = VOCAB.reset_index()
    VOCAB = VOCAB.set_index('term_str')
    VOCAB['term_rank'] = VOCAB['term_rank'] + 1
```

```python
BOW = TOKENS.groupby(bags[bag]+['term_str']).term_str.count().to_frame('n') 
```

```python
TOKENS
```

```python

DTCM = BOW.n.unstack(fill_value=0)
```

```python
tf_method = 'sum'         # sum, max, log, double_norm, raw, binary
tf_norm_k = .5            # only used for double_norm
idf_method = 'standard'   # standard, max, smooth
gradient_cmap = 'YlGnBu'
```

```python
tf = {
    'sum': (DTCM.T / DTCM.T.sum()).T,
}
```

```python
print('TF method:', tf_method)
TF = tf[tf_method]
TF
```

```python
DF = DTCM.astype('bool').sum() 
```

```python
N = DTCM.shape[0]
```

```python
idf = {
    'standard': np.log2(N / DF),
}
```

```python
print('IDF method:', idf_method)
IDF = idf[idf_method]
IDF
```

```python
TFIDF = TF * IDF
```

```python
VOCAB['df'] = DF
VOCAB['idf'] = IDF
```

```python
VOCAB
```

```python
VOCAB['dfidf'] = VOCAB['df'] * VOCAB['idf']
VOCAB.sort_values(by='dfidf', ascending=False).head(21)
```

```python
from sklearn.decomposition import LatentDirichletAllocation as LDA
```

```python
ngram_range = (1, 2)
n_terms = 4000
n_topics = 40
max_iter = 20
n_top_terms = 9
```

```python
lda_engine = LDA(n_components=n_topics, max_iter=max_iter, learning_offset=50., random_state=0)
```

```python
TNAMES = [f"T{str(x).zfill(len(str(n_topics)))}" for x in range(n_topics)]
```

```python
BAG = OHCO[:2]
```

```python
DOCS = TOKENS[TOKENS.pos.str.match(r'^NNS?$')]\
    .groupby(BAG).term_str\
    .apply(lambda x: ' '.join(map(str,x)))\
    .to_frame()\
    .rename(columns={'term_str':'doc_str'})
```

```python
from sklearn.feature_extraction.text import CountVectorizer
count_engine = CountVectorizer(max_features=n_terms, ngram_range=ngram_range, stop_words='english')
count_model = count_engine.fit_transform(DOCS.doc_str)
TERMS = count_engine.get_feature_names_out()
```

```python
lda_model = lda_engine.fit_transform(count_model)
```

```python
THETA = pd.DataFrame(lda_model, index=DOCS.index)
THETA.columns.name = 'topic_id'
THETA.columns = TNAMES
```

```python
THETA.sample(10).T.style.background_gradient(cmap=colors, axis=None)
```

```python
THETA.to_csv("office-THETA.csv")
```

```python
PHI = pd.DataFrame(lda_engine.components_, columns=TERMS, index=TNAMES)
PHI.index.name = 'topic_id'
PHI.columns.name = 'term_str'
```

```python
PHI.to_csv("office-PHI.csv")
```

```python
TOPICS = PHI.stack().groupby('topic_id')\
    .apply(lambda x: ' '.join(x.sort_values(ascending=False).head(n_top_terms).reset_index().term_str))\
    .to_frame('top_terms')
```

```python
TOPICS.style
```

```python
phi_pca = pca_engine.fit_transform(PHI)
```

```python
topic_sizes = THETA.mean(axis=0)
```

```python
import pandas as pd
# Get the dominant topic for each document
dominant_topic = THETA.idxmax(axis=1)  # Series: doc_id → dominant topic

dominant_topic_df = dominant_topic.reset_index()
dominant_topic_df.columns = list(THETA.index.names) + ['dominant_topic']

topic_seasons = dominant_topic_df.groupby('dominant_topic')['season'].agg(lambda x: x.value_counts().idxmax())
```

```python
plot_df = pd.DataFrame({
    'PC1': phi_pca[:, 0],
    'PC2': phi_pca[:, 1],
    'size': topic_sizes,
    'season': topic_seasons  # color by most common season
})
```

```python
import plotly.express as px
import pandas as pd

plot_df = pd.DataFrame({
    'PC1': phi_pca[:, 0],
    'PC2': phi_pca[:, 1],
    'mean_doc_weight': topic_sizes,
    'season': topic_seasons
}).reset_index(drop=True)

fig = px.scatter(
    plot_df,
    x='PC1', y='PC2',
    size='mean_doc_weight',
    color='season',
    hover_name=plot_df.index.astype(str),
    title="Topics in PCA Space (Plotly)",
    labels={'PC1': 'Principal Component 1', 'PC2': 'Principal Component 2'},
    size_max=40,
)

fig.update_traces(marker=dict(opacity=0.8, line=dict(width=1, color='DarkSlateGrey')))
fig.update_layout(height=600, width=800)

fig.show()
```

```python
# word2vec parameters
w2v_params = dict(
    window = 2,
    vector_size = 200,
    min_count = 50, # THIS LIMITS OUR VOCAB
    workers = 4
)
```

```python
import pandas as pd
import numpy as np
from gensim.models import word2vec
from gensim.corpora import Dictionary
from sklearn.manifold import TSNE as tsne
import plotly_express as px
```

```python
docs = TOKENS.dropna(subset=['term_str'])\
    .groupby(BAG)\
    .term_str.apply(lambda  x:  x.tolist())\
    .reset_index()['term_str'].tolist()
docs = [doc for doc in docs if len(doc) > 1] # Lose single word docs
```

```python
vocab = Dictionary(docs)
```

```python
model = word2vec.Word2Vec(docs, **w2v_params)
```

```python
model.wv.vectors
```

```python
def get_vector(row):
    w = row.name
    try:
        vec = model.wv[w]
    except KeyError as e:
        vec = None
    return vec
```

```python
WV = pd.DataFrame(VOCAB.apply(get_vector, axis=1).dropna()).apply(lambda x: pd.Series(x[0]), axis=1)
```

```python
WV.to_csv('office_VOCAB_W2V.csv')
```

```python
def gather(ohco_level):
    global TOKENS
    level_name = OHCO[ohco_level-1].split('_')[0]
    df = TOKENS.groupby(OHCO[:ohco_level])\
        .token_str.apply(lambda x: x.str.cat(sep=' '))\
        .to_frame(f"{level_name}_str")
    return df
```
