```python
# Ensure phi contains numeric data
phi_numeric = phi.apply(pd.to_numeric, errors='coerce')

# Drop columns with all NaNs (in case conversion fails completely on some columns)
phi_numeric = phi_numeric.dropna(axis=1, how='all')

# Now select top N terms
top_n_terms = 30
top_terms = phi_numeric.sum(axis=0).nlargest(top_n_terms).index
phi_top_terms = phi_numeric.loc[:, top_terms]

# Plot heatmap
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
sns.heatmap(phi_top_terms, cmap="viridis", annot=False)
plt.xlabel("Terms")
plt.ylabel("Topics")
plt.title("Topic-Term Heatmap from φ (Phi)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Assume `df` is your DataFrame with columns: 'season', 'episode', 'T00' through 'T39'

# Create an empty list to collect results
results = []

# Loop over all topic columns
topic_cols = [col for col in THETA.columns if col.startswith("T")]

for topic in topic_cols:
    max_idx = THETA[topic].idxmax()
    row = THETA.loc[max_idx, ['season', 'episode']]
    results.append({'topic': topic, 'season': row['season'], 'episode': row['episode'], 'max_value': THETA.loc[max_idx, topic]})

# Convert to a new DataFrame
topic_max_df = pd.DataFrame(results)

# Display result
print(topic_max_df)
```

```python
OHCO = ['season', 'episode', 'scene', 'line', 'sentence', 'token']
raw.rename(columns={'id': 'line'}, inplace=True)
cols = list(raw.columns)
cols.insert(3, cols.pop(cols.index('line')))
raw = raw[cols]
sent_pat = r'[.?!;:]+'
SENTS = raw['line_text'].str.split(sent_pat, expand=True).stack()\
    .to_frame('sentence')
    
cols_to_keep = ['season', 'episode', 'scene', 'line', 'speaker']  # Columns to keep
SENTS[cols_to_keep] = raw.loc[SENTS.index.get_level_values(0), cols_to_keep].values
SENTS.set_index(OHCO[:4], inplace=True)
SENTS
```

```python
import nltk
if True:
    # Return a tokenized copy of text
    # using NLTK's recommended word tokenizer.
    TOKENS = SENTS.sentence\
            .apply(lambda x: pd.Series(nltk.pos_tag(nltk.word_tokenize(x))))\
            .stack()\
            .to_frame('pos_tuple')
```

```python
TOKENS
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re

# Filter for only the four main speakers
main_speakers = ['Michael', 'Jim', 'Pam', 'Dwight']
raw_filtered = raw[raw['speaker'].isin(main_speakers)].copy()
# Initialize a list to store results
data = []

for _, row in TOPICS.iterrows():
    topic_id = row['topic_id']
    
    # Space-separated top terms
    top_words = row['top_terms'].lower().split()
    top_words_set = set(top_words)

    for speaker in main_speakers:
        speaker_lines = raw_filtered[raw_filtered['speaker'] == speaker]['line_text'].str.lower()

        word_count = speaker_lines.apply(
            lambda line: len([word for word in re.findall(r'\b\w+\b', line) if word in top_words_set])
        ).sum()

        data.append({'topic_id': topic_id, 'speaker': speaker, 'count': word_count})

# Create a DataFrame for visualization
counts_df = pd.DataFrame(data)
print(counts_df)

# Plot using seaborn
plt.figure(figsize=(10, 6))
sns.barplot(data=counts_df, x='topic_id', y='count', hue='speaker')
plt.title('Topic Word Usage by Speaker')
plt.xlabel('Topic ID')
plt.ylabel('Count of Topic Words')
plt.legend(title='Speaker')
plt.tight_layout()
plt.show()
```
